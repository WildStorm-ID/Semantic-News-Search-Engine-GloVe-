cek redundant data

```python
import pandas as pd

# 1. Load Dataset
path = "datasets"  # Sesuaikan dengan path foldermu
df = pd.read_csv(path + "/final_merge_dataset.csv")

print("=== 1. INFORMASI UMUM DATASET ===")
print(f"Jumlah baris awal: {len(df)}")
print(df.info())
print("\n" + "="*40 + "\n")

# 2. Cek Missing Values (Nilai Kosong)
print("=== 2. CEK MISSING VALUES (NILAI KOSONG) ===")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0] if missing_data.sum() > 0 else "Tidak ada missing value!")
print("\n" + "="*40 + "\n")

# 3. Cek Duplikat / Redundant Data
print("=== 3. CEK DATA DUPLIKAT ===")
# Cek duplikat berdasarkan judul berita (kolom 'title')
duplicate_titles = df.duplicated(subset=['Judul']).sum()
print(f"Jumlah judul berita duplikat: {duplicate_titles}")

# Cek duplikat persis di seluruh kolom
duplicate_rows = df.duplicated().sum()
print(f"Jumlah baris duplikat persis: {duplicate_rows}")
```

clean-up

```python
# 1. Hapus duplikat berdasarkan kolom judul
df_clean = df.drop_duplicates(subset=['Judul'], keep='first').copy()

# 2. Hapus jika ada baris yang judul beritanya kosong (NaN/null)
df_clean = df_clean.dropna(subset=['Judul'])

# 3. Reset index agar urutan nomor baris rapi
df_clean = df_clean.reset_index(drop=True)

print(f"Jumlah baris awal  : {len(df)}")
print(f"Jumlah baris bersih: {len(df_clean)}")
print(f"Total baris dibuang: {len(df) - len(df_clean)}")

# 4. Simpan ke file CSV BARU
new_csv_path = path + "/indonesian_news_clean.csv"
df_clean.to_csv(new_csv_path, index=False)

print(f"\nSelesai! File baru berhasil disimpan di: {new_csv_path}")
```
